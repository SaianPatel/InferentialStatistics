```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Set a random seed for reproducibility
np.random.seed(42)

# Generate sample data
sample_low = norm.rvs(loc=6, scale=2, size=100)
sample_mid = norm.rvs(loc=10, scale=4, size=100)
sample_mid2 = norm.rvs(loc=12, scale=2, size=100)
sample_high = norm.rvs(loc=18, scale=1, size=100)

# Compute cumulative frequencies
low_sorted = np.sort(sample_low)
mid_sorted = np.sort(sample_mid)
mid_sorted2 = np.sort(sample_mid2)
high_sorted = np.sort(sample_high)

low_cdf = np.arange(1, len(low_sorted) + 1) / len(low_sorted)
mid_cdf = np.arange(1, len(mid_sorted) + 1) / len(mid_sorted)
mid2_cdf = np.arange(1, len(mid_sorted2) + 1) / len(mid_sorted2)
high_cdf = np.arange(1, len(high_sorted) + 1) / len(high_sorted)

# Plot cumulative frequency distributions
plt.figure(figsize=(8, 6))
plt.plot(low_sorted, low_cdf, label='Low Price', linestyle='-', marker='o', markersize=3)
plt.plot(mid_sorted, mid_cdf, label='Mid Price', linestyle='-', marker='s', markersize=3)
plt.plot(mid_sorted2, mid_cdf, label='Mid 2 Price', linestyle='-', marker='s', markersize=3)
plt.plot(high_sorted, high_cdf, label='High Price', linestyle='-', marker='^', markersize=3)

# Labels and title
plt.xlabel('Price')
plt.ylabel('Cumulative Frequency')
plt.title('Cumulative Frequency Distribution of Price Points')
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Set a random seed for reproducibility
np.random.seed(42)

# Generate sample data
sample_low = norm.rvs(loc=7, scale=2, size=100)
sample_mid = norm.rvs(loc=12, scale=4, size=100)
sample_high = norm.rvs(loc=20, scale=4, size=100)

# Plot distributions
plt.figure(figsize=(8, 6))
plt.hist(sample_low, bins=20, alpha=0.6, label='Low Price', density=True)
plt.hist(sample_mid, bins=20, alpha=0.6, label='Mid Price', density=True)
plt.hist(sample_high, bins=20, alpha=0.6, label='High Price', density=True)

# Labels and title
plt.xlabel('Price')
plt.ylabel('Density')
plt.title('Distribution of Price Points')
plt.legend()
plt.grid(True)

# Show plot
plt.show()
```
